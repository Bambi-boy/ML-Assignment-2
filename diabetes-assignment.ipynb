```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# Load the dataset
df = pd.read_csv("diabetes_binary_classification_data.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns=['Diabetes_binary'])
y = df['Diabetes_binary']

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)


print(x_train.head(2), x_test.head(2))
print(y_train.head(2), y_test.head(2))
```

## Feature relationships with the target

The tables below use only the training split. The feature summary compares each feature for `Diabetes_binary` 0 and 1 and ranks features by their absolute correlation with the target. These are descriptive associations, not evidence of causation.

```python
# Summarize target class balance using the training data.
target_summary = y_train.value_counts().sort_index().rename(
    index={0: "No diabetes", 1: "Diabetes"}
).to_frame(name="Count")
target_summary["Percent"] = target_summary["Count"].div(len(y_train)).mul(100)
display(target_summary.round(2))

# Compare average feature values by target class and rank their association.
feature_summary = pd.DataFrame({
    "Mean (no diabetes)": x_train.loc[y_train.eq(0)].mean(),
    "Mean (diabetes)": x_train.loc[y_train.eq(1)].mean(),
})
feature_summary["Difference (diabetes - no diabetes)"] = (
    feature_summary["Mean (diabetes)"] - feature_summary["Mean (no diabetes)"]
)
feature_summary["Correlation with target"] = x_train.corrwith(y_train)
feature_summary["Absolute correlation"] = feature_summary["Correlation with target"].abs()
feature_summary = feature_summary.sort_values(
    "Absolute correlation", ascending=False
)
display(feature_summary.round(3))
```
